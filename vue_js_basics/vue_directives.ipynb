v-bind

The v-bind directive lets us bind an HTML attribute to data in the Vue instance.
This makes it easy to change the attribute value dynamically.

<div v-bind:[attribute]="[vue data]"></div>

```python
<!DOCTYPE html>
<html>
<head>
  <title>'v-bind' Image Source Example</title>
  <style>
    #app {
      border: dashed black 1px;
      width: 250px;
      padding: 0 10px 10px 10px;
    }

    img {
      width: 100%;
    }
  </style>
</head>
<body>

<h1>'v-bind' Image Source Example</h1>

<div id="app">
  <p>The browser finds the 'src' attribute value from the Vue instance with the use of 'v-bind'.</p>
  <img v-bind:src="url">
</div>

<script src="https://unpkg.com/vue@3/dist/vue.global.js"></script>
<script>
  const app = Vue.createApp({
   data() {
    return {
      url: "https://miro.medium.com/v2/resize:fit:4800/format:webp/1*SdXRP8f2Lhin89Tht_GRIA.jpeg"
    }
   }
  })
  app.mount('#app')
</script>

</body>
</html>
```
