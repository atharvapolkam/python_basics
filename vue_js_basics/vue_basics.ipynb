Vue directives are special HTML attributes with the prefix v- that give the HTML tag extra functionality.

Vue directives connect to the Vue instance to create dynamic and reactive user interfaces.

The most important Vue.js concepts to learn

1. Template syntax and directives
Essential
Learn how Vue displays dynamic data and controls what appears on screen.
- {{ }} — display values
- v-if / v-else — conditional rendering
- v-for — display lists
- v-bind (:) — bind HTML attributes
- v-on (@) — handle events
Why: Used in nearly every Vue component to display data, alerts, tables, and buttons.




2. Reactivity: ref() and reactive()
Essential
Learn how Vue updates the interface automatically when data changes.
import { ref } from 'vue'

const count = ref(0)

function increment() {
  count.value++
}


Why: Useful for counters, form inputs, loading states, alert counts, and live dashboard values.
In JavaScript inside Vue components, use count.value; in templates, Vue unwraps refs automatically.




3. Components and props
Learn to divide your UI into reusable pieces and pass data between them.
- Components: App.vue, Navbar.vue, AlertTable.vue
- Props: parent component passes data to child
- Emits: child notifies parent about an event
Why: Avoids repeating code and makes large applications easier to maintain.




4. Event handling and forms
Learn @click, @submit, and v-model.
Why: Every interactive application needs buttons, search fields, dropdowns, and forms.
Practice creating a form that submits camera configuration to a backend API.




5. Computed properties and watchers
- computed() — derive values from existing state.
- watch() — respond when a particular value changes.
Why: Computed properties are excellent for filtering and counting alerts. Watchers are useful when a change must trigger an API request or another side effect.




6. API integration
Learn fetch() or Axios, async/await, JSON, HTTP methods, and error handling.
Why: Vue handles the frontend; your Django or FastAPI backend can handle business logic, authentication, databases, and AI inference.
Practice fetching camera data, submitting configurations, and displaying detection results.




7. Lifecycle hooks
Start with onMounted() and onUnmounted().
Why: Fetch initial data when a page loads, and clean up timers or event listeners when a component is removed.




8. Vue Router
Learn routes, route parameters, and navigation guards.
Why: Needed when your application has separate pages such as Login, Dashboard, Cameras, and Alert History.




9. State management with Pinia
Learn this after understanding ref(), components, and props.
Why: Shares application-wide data such as the logged-in user, selected site, and alert filters across multiple pages.
