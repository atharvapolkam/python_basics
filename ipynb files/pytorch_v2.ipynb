
# PyTorch Workflow - Complete Beginner Notes

## Machine Learning Workflow

Data
↓
Train Model
↓
Learn Pattern
↓
Predict

---

## Step 1: Create Data

Assume:

y = 0.7x + 0.3

Example:

x = 2

y = 0.7(2) + 0.3

y = 1.7

Why create fake data?

Because we already know the correct answer and can check whether the model learns it.

---

## Step 2: Train/Test Split

100 samples

80 -> Training

20 -> Testing

Training data teaches the model.

Testing data checks whether the model learned correctly.

---

## Step 3: Build Model

Model:

y = wx + b

w = weight

b = bias

Initially:

weight and bias are random.

Example:

weight = 0.12

bias = 0.89

Training improves them.

---

## Step 4: Forward Pass

Forward pass means making a prediction.

Example:

x = 5

Prediction:

y = 0.12×5 + 0.89

= 1.49

---

## Step 5: Loss Function

Prediction = 1.49

Actual = 3.8

Loss measures how wrong the prediction is.

Large Loss -> Bad

Small Loss -> Good

---

## Step 6: Optimizer

Optimizer updates weights.

Example:

weight = 0.12

Loss is high.

Optimizer changes:

weight = 0.20

Then:

weight = 0.35

Then:

weight = 0.50

Loss gradually decreases.

---

## Step 7: Training Loop

Predict
↓
Calculate Loss
↓
Update Weights
↓
Repeat

Example:

Epoch 1 -> Loss 10

Epoch 2 -> Loss 7

Epoch 3 -> Loss 4

Epoch 4 -> Loss 2

Epoch 5 -> Loss 0.5

---

## What is an Epoch?

One complete pass through all training data.

100 images

Model sees all 100 images once

= 1 Epoch

---

## Final Goal

Real Formula:

y = 0.7x + 0.3

Model Starts:

y = 0.12x + 0.89

After Training:

y = 0.699x + 0.301

The model discovered the hidden pattern.

# computation graph 

What PyTorch secretly stores

It creates a graph.

          x
          |
       ×3
          |
          y
          |
       +4
          |
          z
          |
      square
          |
        loss

Every operation becomes a node.

PyTorch remembers

multiplication
addition
square

This is called the Computation Graph.


What happens when you call
loss.backward()

PyTorch starts from the end.

loss
↑
z
↑
y
↑
x

It walks backwards through the graph.

This is called Backpropagation.


w.grad is told network how much value should be adjusted and in which direction


```python
import torch

# Input and target values
x = torch.tensor(2.0)
y_true = torch.tensor(10.0)

# Weight to learn
w = torch.tensor(1.0, requires_grad=True)

# Learning rate
lr = 0.1

# Forward pass
y_pred = w * x
loss = (y_pred - y_true) ** 2

# Backpropagation: calculate gradient
loss.backward()

print("Before update")
print("w:", w.item())
print("loss:", loss.item())
print("gradient:", w.grad.item())

# Update the weight
with torch.no_grad():
    w -= lr * w.grad

# Clear the previous gradient
w.grad.zero_()

print("\nAfter update")
print("w:", w.item())

# Check the new loss
y_pred = w * x
loss = (y_pred - y_true) ** 2
print("new loss:", loss.item())
```

```python
######training######

import torch

# Training data: y = 2x
x = torch.tensor([1.0, 2.0, 3.0, 4.0])
y_true = torch.tensor([2.0, 4.0, 6.0, 8.0])

# Initial weight and learning rate
w = torch.tensor(0.0, requires_grad=True)
lr = 0.1
epochs = 20

for epoch in range(epochs):
    # Step 1: prediction
    y_pred = w * x

    # Step 2: mean squared error loss
    loss = ((y_pred - y_true) ** 2).mean()

    # Step 3: calculate gradient
    loss.backward()

    # Step 4: update weight
    with torch.no_grad():
        w -= lr * w.grad

    # Clear gradients before the next epoch
    w.grad.zero_()

    print(f"Epoch {epoch + 1}: w={w.item():.4f}, loss={loss.item():.4f}")

print("\nFinal weight:", w.item())

# Test prediction
test_x = torch.tensor(5.0)
print(f"Prediction for x={test_x.item()}: {w * test_x:.4f}")
```

```python
# CNN

# CONVOLUTIONAL MEANS KERNAL SLIDES OVER THE FULL IMAGE

#KERNALS ARE MULTIPLE LAYERS
# 1ST LAYER WILL DETECT LINES , EDGES
# 2ND WILL DETECT CURVERS , CORNERS
# 3RD SHAPES


import torch
import torch.nn as nn

conv = nn.Conv2d(
    in_channels=3,     # RGB image: Red, Green, Blue
    out_channels=16,   # learn 16 different filters/features
    kernel_size=3,     # each filter is 3 × 3
    stride=1,          # move 1 pixel at a time
    padding=1,         # add 1-pixel border
    bias=True
)

image = torch.randn(8, 3, 32, 32)
output = conv(image)

print(output.shape)  # torch.Size([8, 16, 32, 32])

'''
| Parameter | Meaning |
|---|---|
| `in_channels` | Number of channels in the input image. `1` for grayscale, `3` for RGB. |
| `out_channels` | Number of filters the layer learns; also the output-channel count. |
| `kernel_size` | Filter size, e.g. `3` means a `3 × 3` filter. |
| `stride` | How far the filter moves on each step. Larger stride makes output smaller. |
| `padding` | Extra pixels around the image border. |
| `bias` | One learned extra value for each output channel. |
'''
```

```python
#training of CNN
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

# Use GPU if available; otherwise use CPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# Download and prepare MNIST dataset
transform = transforms.ToTensor()

train_dataset = datasets.MNIST(
    root="./data",
    train=True,
    download=True,
    transform=transform
)

test_dataset = datasets.MNIST(
    root="./data",
    train=False,
    download=True,
    transform=transform
)

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False)


# CNN model
class CNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(1, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),  # 28x28 -> 14x14

            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2)   # 14x14 -> 7x7
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(32 * 7 * 7, 128),
            nn.ReLU(),
            nn.Linear(128, 10)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x


model = CNN().to(device)

# Loss function and optimizer
loss_fn = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Training
epochs = 3

for epoch in range(epochs):
    model.train()
    total_loss = 0

    for images, labels in train_loader:
        images = images.to(device)
        labels = labels.to(device)

        # Forward pass
        outputs = model(images)
        loss = loss_fn(outputs, labels)

        # Backward pass and weight update
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    average_loss = total_loss / len(train_loader)
    print(f"Epoch {epoch + 1}/{epochs}, Loss: {average_loss:.4f}")


# Testing / accuracy
model.eval()
correct = 0
total = 0

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)
        labels = labels.to(device)

        outputs = model(images)
        predictions = torch.argmax(outputs, dim=1)

        total += labels.size(0)
        correct += (predictions == labels).sum().item()

accuracy = 100 * correct / total
print(f"Test Accuracy: {accuracy:.2f}%")
```
