# FastAPI Notes

## 1. What is a Server?

A server is a program that waits for requests from clients and sends responses.

Client → Request → Server  
Client ← Response ← Server

Example:

Browser → GET /users → FastAPI Server  
Browser ← JSON Response ← FastAPI Server


## 2. What is a Client?

A client is a program that sends requests to a server.

Examples:

- Browser
- Mobile application
- Frontend JavaScript
- Python program
- Postman
- Another backend service

Example:

Python Client → FastAPI Server → JSON Response


## 3. What is HTTP?

HTTP stands for HyperText Transfer Protocol.

It defines how clients and servers communicate.

An HTTP Request contains:

- Method
- URL
- Headers
- Body

An HTTP Response contains:

- Status Code
- Headers
- Body
