 [1,2,3],
 [4,5,6]
]

# 6. Dot Product

Vectors:

a = [1,2,3]

b = [4,5,6]

Multiply matching positions:

1×4 = 4
2×5 = 10
3×6 = 18

Add results:

4 + 10 + 18 = 32

Dot Product = 32

[0.5, 0.3, 0.2]

weight×0.3 +

# 8. Matrix Multiplication - Easy Way

Think of matrix multiplication as:

Input Information
        ↓
Combine Information
        ↓
Output Information

Rule 1:
Inner dimensions must match.

(2,3) × (3,2) -> Valid

Rule 2:
Output shape uses outer dimensions.

(A,B) × (B,C) = (A,C)


# 9. Matrix Multiplication Example

A

[
 [1,2,3],
 [4,5,6]
]

B

[
 [7,8],
 [9,10],
 [11,12]
]

Step 1:

Take first row from A

[1,2,3]

Take first column from B

[7,9,11]

Dot Product:

1×7 + 2×9 + 3×11

= 7 + 18 + 33

= 58

First value = 58

Repeat for every row and column.

Final Result:

[
 [58,64],
 [139,154]
]


# 10. Real Life Meaning of Matrix Multiplication

Student Scores

[80,70,90]

Weights

[0.5,0.3,0.2]

Result

80×0.5 + 70×0.3 + 90×0.2

= 79

What happened?

Three numbers were converted into one meaningful score.

Matrix multiplication transforms information.


# 11. Images as Tensors

Image Shape:

(224,224,3)

Meaning:

224 rows (height)
224 columns (width)
3 channels (RGB)

Each pixel:

[R,G,B]

Examples:

[255,0,0] = Red
[0,255,0] = Green
[0,0,255] = Blue


# 12. Batch of Images

Shape:

(32,224,224,3)

Meaning:

32 images

Each image:
224 rows
224 columns
3 channels

Dimension Meaning:

0 -> Batch Size
