```python
#gaussian blur

1)smoothening of image soften the image

syntax: cv2.GaussianBlur(src, ksize, sigmaX)

# what is kernal size and sigmaX

ksize: Size of the kernel (should be odd and positive) # moves along the image and applies the blur effect to the neighboring pixels defined by the kernel size window
sigmaX: Standard deviation in the X direction
```

```python
#median blur

median blur preserves edges while removing noise. 
It replaces each pixel's value with the median value of the neighboring pixels
defined by the kernel size.
```
